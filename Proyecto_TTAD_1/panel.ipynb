```python
import json

# Cargamos los datos del archivo JSON
with open("data.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

# Obtenemos la lista de ventas
ventas = datos["ventas"]

print(ventas)
```

```python
# INGRESOS DE CADA VENTA
for venta in ventas:

    ingresos = venta["cantidad"] * venta["precio_unitario"]

    print(venta["producto"], ingresos)


  
        
```

```python
# EL PRODUCTO CON MÁS INGRESOS
ingresos_productos = {}
for venta in ventas:

    producto = venta["producto"]
    ingresos = venta["cantidad"] * venta["precio_unitario"]

    if producto in ingresos_productos:
        ingresos_productos[producto] = ingresos_productos[producto] + ingresos
    else:
        ingresos_productos[producto] = ingresos

ingreso_maximo = max(ingresos_productos.values())

for producto in ingresos_productos:

    if ingresos_productos[producto] == ingreso_maximo:
        print("Producto:", producto)
        print("Ingresos:", ingreso_maximo, "€")
```

```python
# FILTRAR LAS VENTAS POR UN VENDEDOR CONCRETO
vendedor_buscado = "Ana"
for venta in ventas:
     if venta["vendedor"] == vendedor_buscado:

        print("ID:", venta["id"])
        print("Producto:", venta["producto"])
        print("Mes:", venta["mes"])
        print("Cantidad:", venta["cantidad"])
        print("Precio unitario:", venta["precio_unitario"])
        print("-------------------------")
```

```python
# CALCULAR EL TICKET MEDIO
ingresos_totales = 0
numero_ventas = len(ventas)

for venta in ventas:

    ingresos = venta["cantidad"] * venta["precio_unitario"]

    ingresos_totales = ingresos_totales + ingresos

ticket_medio = ingresos_totales / numero_ventas

print("Ingresos totales:", ingresos_totales, "€")
print("Número de ventas:", numero_ventas)
print("Ticket medio:", ticket_medio, "€")
```
